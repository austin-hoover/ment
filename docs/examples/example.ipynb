# Example notebook

This notebook demonstrates a small MENT-style reconstruction workflow and is rendered directly into the HTML documentation.

## Imports

The notebook can contain regular Markdown, equations, code cells, and plots.

```python
import matplotlib.pyplot as plt
import numpy as np
import torch

import ment
```

## Generate a simple 2D distribution

Here we generate a correlated Gaussian cloud as a lightweight example dataset.

```python
rng = np.random.default_rng(0)

mean = np.array([0.0, 0.0])
cov = np.array([
    [1.0, 0.7],
    [0.7, 1.5],
])

x = rng.multivariate_normal(mean, cov, size=10_000)
x = torch.tensor(x, dtype=torch.float32)

x.shape
```

## Visualize the source distribution

```python
fig, ax = plt.subplots(figsize=(4, 4))
ax.hist2d(x[:, 0].numpy(), x[:, 1].numpy(), bins=80)
ax.set_xlabel("x")
ax.set_ylabel("y")
ax.set_title("Example 2D distribution")
plt.show()
```

## Compute a covariance matrix

MENT includes utilities for reconstruction workflows, and covariance fitting is often used as a preprocessing step. This minimal notebook keeps the example lightweight for documentation builds.

```python
cov_matrix = torch.cov(x.T)
cov_matrix
```
